```python
import pandas as pd
```

```python
marks=[25,67,89,24,99,15,33,62,77,92]
df=pd.DataFrame({"Marks":marks})
#Calculate Quantile
Q1= df["Marks"].quantile(0.25)
Q2= df["Marks"].quantile(0.50)
Q3= df["Marks"].quantile(0.75)

#Calculate IQR
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("Q1:",Q1)
print("Q2:",Q2)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower bound:",lower_bound)
print("Upper bound:",upper_bound)
outliers=df[(df["Marks"]<lower_bound) | (df["Marks"]>upper_bound)]
print("Outliers:",outliers)
```

```python
values=[48,49,50,51,51,52,52,52,52,53,53,53,53,53,54,54,55,55,55,100]
mean=sum(values)/len(values)
std=pd.Series(values).std(ddof=0)
print("Mean:",round(mean,2))
print("Standard deviation:",round(std,2))

#Detect outliers
for x in values:
  z=(x-mean)/std
  if abs(z)>3:
    print("Outlier:",x)
    print("Z-score:",round(z,2))
```
